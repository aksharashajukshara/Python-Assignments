```python
import numpy as np
```

```python
# 1. Create a 1D NumPy Array:

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# 2. Inspection and Properties:

# Shape,data type,number of elements

print("Shape:",temperatures_w1.shape)
print("Data type:",temperatures_w1.dtype)
print("Number of elements:",temperatures_w1.size)
```

```python
# 3. Array Operations:

#Celsius to Fahrenheit 

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
Fahrenheit = (temperatures_w1* 9/5) + 32
print(Fahrenheit)

# Maximum,Minimum,Mean

print("Maximum Temperature:",temperatures_w1.max())
print("Minimum Temperature:",temperatures_w1.min())
print("Mean Temperature:",temperatures_w1.mean())
```

```python
# 4. Array Slicing and Indexing:

print("First 3 Days:",temperatures_w1[0:3])
print("Last 2 Days:",temperatures_w1[-2:])
print("Middle 3 Days:",temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:

temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],              #Week1
 [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])                                   #Week2
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array:

# Shape,data type,number of elements
print("Shape:",temperatures.shape)
print("Data type:",temperatures.dtype)
print("Number of elements:",temperatures.size)

# Temperature 
print("Week1 Temperature:",temperatures[0])
print("Week2 Temperature:",temperatures[1])
print("Weekend Temperature for both weeks",temperatures[:,-2:])
```

```python
## CREATING PANDAS
import pandas as pd
```

```python
marks=pd.Series([95, 92, 89, 85, 80],
                index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"])
print(marks)
```

```python
# 2. Indexing and Slicing:
# Mark at 1 position
print("marks of 1 position:",marks.iloc[0])
```

```python
# mark for the top3
print("marks of top 3:",marks.loc[["Rank1", "Rank2", "Rank3"]])
```

```python
print(marks[marks>90])
```

```python
# 3. Manipulating Series:
# Modify the mark of the student
marks.loc["Rank1"]=100
print(marks)
```

```python
# Remove the entry of last row
marks=marks.drop("Rank5")
print(marks)
```

```python
# Compute the CGPA
CGPA=marks/10
print(CGPA)
```

```python
# Pandas DataFrame:

#1. Creating Pandas DataFrame:

transactions={
    "TransactionID":[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory":["Electronics", "Clothing", "Electronics", "Furniture",
                    "Clothing", "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"],
    "Region":["North", "South", "North", "East", "West", "North", "East", "West",
                    "South", "North"],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions=pd.DataFrame(transactions)
print(transactions)
```

```python
# 2. Data Exploration:

# head, tail, shape, column names, and data types.

print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print(transactions.shape)
```

```python
print(transactions.columns)
```

```python
print(transactions.dtypes)
```

```python
##  Display only the 'ProductCategory' and 'Amount' columns.

print(transactions[["ProductCategory", "Amount"]])
```

```python
## Retrieve

print(transactions.iloc[:, -3:])
```

```python
## Filter

print(transactions[(transactions["Region"] == "North") & (transactions["Amount"] > 200)])
```

```python
## Values Count

transactions["ProductCategory"].value_counts()
```

```python
## Unique Values

print(transactions["Region"].unique())
```

```python
## GroupBy

print(transactions.groupby("Region")["Amount"].mean())
```

```python
# 3. Manipulating the DataFrame:

## Modify
transactions.loc[transactions["TransactionID"]==
    102, "Amount"]=165
print(transactions)
```

```python
## Adding Column
transactions["Discount"]=transactions["Amount"]*0.10
print(transactions)
```

```python
## Removing the row
transactions = transactions.drop(index=8)
print(transactions)
```

```python
## Deleting Discount Column
transactions=transactions.drop(columns=["Discount"])
print(transactions)
```
