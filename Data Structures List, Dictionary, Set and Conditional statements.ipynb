```python
## 1. List (Creation, Modification and Access):
age_list=[24, 25, 26, 27, 28]
name_list=["Akshara","Anagha","Arannya","Athulya","Thasni"]
```

```python
## 2. List Operations / Modifications:
age_list=[24, 25, 26, 27, 28]
name_list=["Akshara","Anagha","Arannya","Athulya","Thasni"];
## a)
name_list.append("Yazhini")
print(name_list)
## b)
age_list.insert(2, 30)
print(age_list)
## c)
name_list.remove("Yazhini")
print(name_list)
## d)
age_list.pop()
print(age_list)
## e)
age_list.extend([29, 30, 26])
print(age_list)
## f)
age_list.sort()
print(age_list)
## g)
## Maximum age
print(max(age_list))
## Minimum age
print(min(age_list))
## Sum of all age
print(sum(age_list))
```

```python
## 3. Accessing List Elements:
name_list=["Akshara","Anagha","Arannya","Athulya","Thasni"]
## a)
print(name_list[0])
## b)
print(name_list[-1])
## c)
print(name_list[2:4])
## d)
print(name_list[::-1])
```

```python
## Dictionary (Creation, Modification and Access):
## a)
students_mark={"Akshara":98,
"Anagha":94,
"Arannya":89,
"Athulya":84,
"Thasni":70}
## b)
print(students_mark["Anagha"])
## c)
students_mark["Janani"]=80
print(students_mark)
## d)
students_mark.update({"Thasni":82})
print(students_mark)
## e)
print(students_mark.keys())
print(students_mark.values())
```

```python
##  Sets (Operations):
## a)
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
## Explanation
# 1. A Set is an unordered collection of unique elements.
# 2. Mutable
# 3. No duplicate values
# 4. Cannot access items using indexing



```

```python
## b)
my_set[4]="s"
print(my_set)
## Explanation
# 1. Sets cannot be indexed because they are unordered.
```

```python
## c) and d)
set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print("Union:",set1.union(set2))
print("Intersection:",set1.intersection(set2))
```

```python
## Operators & Conditional Statements :(IF, ELIF, ELSE)
## 1.
score = int(input("Enter a score (0-10): "))

if score < 0 or score > 10:
    print("Invalid score")
elif score > 7:
    print("Above Average")
    print("Excellent work! Keep it up.")
elif score >= 4:
    print("Average")
    print("You are doing well. Keep improving.")
else:
    print("Below Average")
    print("You need to improve your performance. Consistent practice will lead to better results.")
```
