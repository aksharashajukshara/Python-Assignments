```python
## 1. String Concatenation:
string1="Hello"
Enter_your_Name="Akshara"
print(string1+" "+EnteryourName);
```

```python
string1 = "Hello "
Enter_your_Name = "Akshara"
string3 = ", welcome to Python programming"
print(string1 + Enter_your_Name + string3)
```

```python
## 2. String Slicing and Indexing:
string1 = "Hello "
Enter_your_Name = "Akshara"
string3 = ", welcome to Python programming"
text= string1 + Enter_your_Name + string3
print(text[0]);
print(text[-1]);
print(text[0:5]);
print(text[-11:]);
print(text[::-1]);
print(text[26:32]);
```

```python
## 3. String Methods:
strM = "Python beginner tutorial"
print(strM.upper());
print(strM.lower());
print(strM.capitalize());
print(strM.count("t"));
print(strM.replace("Python","Machine Learning"));
```

```python
## 4. Tuples:
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
t_combine=(tuple1+tuple2)
print(t_combine);
print(t_combine*3);
print(t_combine[2]);
print(t_combine[:3]);
print(t_combine[-3:]);
```
